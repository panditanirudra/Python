```python
print("Welcome to the Library Management System")

library = {
    "Python Basics": {"author": "Test", "available": True, "issued_to": None},
    "Data Science": {"author": "Test1", "available": False, "issued_to": "Alice"}
}

credentials = {"librarian": "lib123", "admin": "admin123"}


def display_books():
    print("\n--- Library Books ---")
    for title, details in library.items():
        status = "Available" if details["available"] else f"Issued to {details['issued_to']}"
        print(f"Title: {title}, Author: {details['author']}, Status: {status}")
    print()


def add_book():
    title = input("Enter book title: ")
    author = input("Enter book author: ")
    if title in library:
        print("Book already exists.")
    else:
        library[title] = {"author": author, "available": True, "issued_to": None}
        print("Book added successfully!")


def issue_book():
    title = input("Enter the book title to issue: ")
    if title in library:
        if library[title]["available"]:
            name = input("Enter your name: ")
            library[title]["available"] = False
            library[title]["issued_to"] = name
            print(f"The book '{title}' has been issued to {name}.")
        else:
            print(f"The book '{title}' is already issued to {library[title]['issued_to']}.")
    else:
        print("Book not found.")


def return_book():
    title = input("Enter the book title to return: ")
    if title in library:
        if not library[title]["available"]:
            print(f"Book returned by {library[title]['issued_to']}.")
            library[title]["available"] = True
            library[title]["issued_to"] = None
        else:
            print("This book is not issued.")
    else:
        print("Book not found.")


def view_issued_books():
    print("\n--- Issued Books ---")
    found = False
    for title, details in library.items():
        if not details["available"]:
            print(f"Title: {title}, Issued To: {details['issued_to']}")
            found = True
    if not found:
        print("No books are currently issued.")
    print()


def login():
    attempts = 3
    while attempts > 0:
        username = input("Enter username: ")
        password = input("Enter password: ")
        if username in credentials and credentials[username] == password:
            print(f"Welcome {username}!\n")
            return True
        else:
            attempts -= 1
            print(f"Invalid credentials. Attempts left: {attempts}")
    print("Access denied.")
    return False


def main():
    if not login():
        return

    while True:
        print("==== Library Menu ====")
        print("1. View Books")
        print("2. Add a Book")
        print("3. Issue a Book")
        print("4. Return a Book")
        print("5. View Issued Books")
        print("6. Exit")
        choice = input("Enter your choice (1-6): ")

        if choice == "1":
            display_books()
        elif choice == "2":
            add_book()
        elif choice == "3":
            issue_book()
        elif choice == "4":
            return_book()
        elif choice == "5":
            view_issued_books()
        elif choice == "6":
            print("Thank you for using the Library Management System.")
            break
        else:
            print("Invalid choice. Please try again.\n")


if __name__ == "__main__":
    main()
```
