```python
import numpy as np

# Q1. Basic arithmetic operations

arr1 = np.array([1, 2, 3, 0])
arr2 = np.array([4, 1, 0, 6])

print("Addition:", arr1 + arr2)
print("Subtraction:", arr1 - arr2)
print("Multiplication:", arr1 * arr2)
# Handle division by zero safely
with np.errstate(divide='ignore', invalid='ignore'):
    division_result = np.true_divide(arr1, arr2)
    division_result[~np.isfinite(division_result)] = np.nan
print("Division:", division_result)
```

```python
# Q2. 1-D and 2-D random arrays

arr1d = np.random.randint(1, 101, 5)
arr2d = np.random.randint(1, 101, (4, 5))

print("\n1-D array:", arr1d)
print("2-D array:\n", arr2d)

print("Addition:\n", arr2d + arr1d)
print("Subtraction:\n", arr2d - arr1d)
```

```python
# Q3. 3-D array indexing/slicing
# -------------------------------
arr3d = np.random.randint(1, 101, (3, 3, 3))
print("\n3-D array:\n", arr3d)

print("Third column of second 2-D array:", arr3d[1, :, 2])
print("First and third row of first 2-D array:\n", arr3d[0, [0, 2], :])
print("Intersection of first row and third column in third 2-D array:", arr3d[2, 0, 2])
```

```python
# Q4. 2-D array even/odd/sum/reshape

arr2d_q4 = np.random.randint(1, 101, (5, 4))
print("\nQ4 Array:\n", arr2d_q4)

print("Even numbers:", arr2d_q4[arr2d_q4 % 2 == 0])
print("Odd numbers:", arr2d_q4[arr2d_q4 % 2 != 0])
print("Row-wise sum:", arr2d_q4.sum(axis=1))
print("Column-wise sum:", arr2d_q4.sum(axis=0))

arr3d_q4 = arr2d_q4.reshape(2, 2, 5)
print("Reshaped 3-D array (2x2x5):\n", arr3d_q4)
```

```python
import numpy as np
# Q5. Row/column swaps, replacement, reshape

arr2d_q5 = np.random.randint(1, 101, (5, 4))
print("\nQ5 Original Array:\n", arr2d_q5)

# Swap first and third row
arr2d_q5[[0, 2]] = arr2d_q5[[2, 0]]

# Swap second and fourth column
arr2d_q5[:, [1, 3]] = arr2d_q5[:, [3, 1]]

# Replace values less than 50 with zero
arr2d_q5[arr2d_q5 < 50] = 0

# Convert to 1-D array
arr1d_q5 = arr2d_q5.flatten()

print("Modified Array:\n", arr2d_q5)
print("Flattened 1-D Array:", arr1d_q5)

```

```python
import numpy as np

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

# Stacking (creates a new axis)
print("\nVertical Stack:\n", np.vstack((a, b)))
print("Horizontal Stack:\n", np.hstack((a, b)))

# Concatenation (joins along an existing axis)
print("Concatenate along axis 0:\n", np.concatenate((a, b), axis=0))
print("Concatenate along axis 1:\n", np.concatenate((a, b), axis=1))
```
